```python
# ==============================
# 1. Conectar Google Drive
# ==============================

from google.colab import drive
drive.mount('/content/drive')


# ==============================
# 2. Importar librerías
# ==============================

import pandas as pd
from scipy.stats import chi2_contingency


# ==============================
# 3. Cargar los archivos
# Cambia la ruta según donde tengas tus archivos en Drive
# ==============================

real = pd.read_excel('/content/drive/MyDrive/Colab Notebooks/DataArSGamExtend/realData.xlsx')

synthetic = pd.read_excel('/content/drive/MyDrive/Colab Notebooks/DataArSGamExtend/syntheticData.xlsx')


# ==============================
# 4. Verificar columnas
# ==============================

print("Columnas dataset real:")
print(real.columns)

print("\nColumnas dataset sintético:")
print(synthetic.columns)


# ==============================
# 5. Agregar columna de origen
# ==============================

real["Dataset"] = "Real"
synthetic["Dataset"] = "Synthetic"


# ==============================
# 6. Unir datasets
# ==============================

df = pd.concat([real, synthetic], ignore_index=True)


# ==============================
# 7. Función para Chi-cuadrado
# ==============================

def chi_square_analysis(data, variable):
    print("\n" + "="*60)
    print(f"VARIABLE ANALIZADA: {variable}")
    print("="*60)

    # Tabla de contingencia
    table = pd.crosstab(data[variable], data["Dataset"])

    print("\nTabla de contingencia:")
    print(table)

    # Prueba Chi-cuadrado
    chi2, p, dof, expected = chi2_contingency(table)

    expected_df = pd.DataFrame(
        expected,
        index=table.index,
        columns=table.columns
    )

    print("\nFrecuencias esperadas:")
    print(expected_df)

    print("\nResultados Chi-cuadrado:")
    print(f"Chi-square = {chi2:.4f}")
    print(f"p-value = {p:.4f}")
    print(f"Degrees of freedom = {dof}")

    # Verificación de frecuencias esperadas menores a 5
    low_expected = (expected < 5).sum()

    print(f"\nNúmero de celdas con frecuencia esperada menor a 5: {low_expected}")

    if low_expected > 0:
        print("Advertencia: La prueba debe interpretarse con cautela por frecuencias esperadas bajas.")

    # Interpretación
    if p > 0.05:
        print("\nInterpretación:")
        print("No se encontraron diferencias estadísticamente significativas entre las distribuciones real y sintética.")
    else:
        print("\nInterpretación:")
        print("Se encontraron diferencias estadísticamente significativas entre las distribuciones real y sintética.")


# ==============================
# 8. Ejecutar análisis
# Asegúrate de que los nombres de las columnas coincidan exactamente
# ==============================

chi_square_analysis(df, "Emotion")
chi_square_analysis(df, "Core Drive Octalysis")
chi_square_analysis(df, "Motivation")
```
