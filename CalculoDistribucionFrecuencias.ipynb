```python
import pandas as pd

# Cargar datasets
real = pd.read_excel("realData.xlsx")
synthetic = pd.read_excel("syntheticData.xlsx")

def frequency_distribution(real_df, synthetic_df, variable):

    real_abs = real_df[variable].value_counts()
    synthetic_abs = synthetic_df[variable].value_counts()

    categories = sorted(
        set(real_abs.index).union(set(synthetic_abs.index))
    )

    results = []

    for cat in categories:

        real_freq = real_abs.get(cat, 0)
        synthetic_freq = synthetic_abs.get(cat, 0)

        real_rel = real_freq / len(real_df)
        synthetic_rel = synthetic_freq / len(synthetic_df)

        results.append([
            cat,
            real_freq,
            round(real_rel*100,2),
            synthetic_freq,
            round(synthetic_rel*100,2)
        ])

    table = pd.DataFrame(
        results,
        columns=[
            "Category",
            "Real Frequency",
            "Real %",
            "Synthetic Frequency",
            "Synthetic %"
        ]
    )

    return table

# Emotion
emotion_table = frequency_distribution(
    real,
    synthetic,
    "Emotion"
)

print(emotion_table)

# Motivation
motivation_table = frequency_distribution(
    real,
    synthetic,
    "Motivation"
)

print(motivation_table)

# Core Drive
core_table = frequency_distribution(
    real,
    synthetic,
    "Core Drive Octalysis"
)

print(core_table)
```
